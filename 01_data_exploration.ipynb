## "In this notebook , 
I explore the StatsBomb open dataset to identify available competitions and matches for tactical analysis."

```python
from statsbombpy import sb
import pandas as pd

# 1. Fetch all available free competitions
print("Connecting to StatsBomb API... Fetching competitions.")
competitions = sb.competitions()

# Define the primary columns for identification
target_columns = ['competition_name', 'season_name', 'competition_id', 'season_id']

# Display the top 10 available competitions
print("\n--- Available Competitions ---")
print(competitions[target_columns].head(10))

# 2. Select FIFA World Cup 2022 (Comp ID: 43, Season ID: 106)
# This dataset is highly detailed and perfect for tactical modeling
COMP_ID = 43
SEASON_ID = 106

print(f"\nFetching matches for Competition ID: {COMP_ID}, Season ID: {SEASON_ID}...")
wc_matches = sb.matches(competition_id=COMP_ID, season_id=SEASON_ID)

# Display the first 10 matches to select a specific Match ID
print("\n--- Match List (World Cup 2022) ---")
print(wc_matches[['match_id', 'home_team', 'away_team', 'match_date']].head(10))
```
